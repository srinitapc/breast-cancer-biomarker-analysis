```python
import pandas as pd
import matplotlib.pyplot as plt

dataset = pd.read_csv("/Breast Cancer Wisconsin Breast Cancer (Diagnostic) Data Set.csv", index_col=0)
dataset
malignant_df = dataset[dataset['diagnosis']== 'M']
malignant_df
benign_df = dataset[dataset['diagnosis']== 'B']
benign_df
# Malignant means
m_area_mean = malignant_df['area_mean'].mean()
m_radius_mean = malignant_df['radius_mean'].mean()
m_concavity_mean = malignant_df['concavity_mean'].mean()
#Benign Means
b_area_mean = benign_df['area_mean'].mean()
b_radius_mean = benign_df['radius_mean'].mean()
b_concavity_mean = benign_df['concavity_mean'].mean()

#Mean comparisons
print(m_area_mean/b_area_mean)
print(m_radius_mean/b_radius_mean)
print(m_concavity_mean/b_concavity_mean)
  # Based on this comparison concavity signaled the most significant difference
# between begnign and malignant cells

#Histograms
  #Area Mean
plt.hist(benign_df["area_mean"],bins = 40, alpha=0.5, label='Benign', color='blue')
plt.hist(malignant_df["area_mean"], bins = 40, alpha=0.5, label='Malignant', color='red')
plt.legend(loc='upper right')
plt.title('Distribution of Area Mean: Malignant vs Benign')
plt.xlabel('Area Mean (Variable)')
plt.ylabel('Count (Observations)')

plt.show()
  # Radius Mean
plt.hist(benign_df["radius_mean"], bins = 40, alpha=0.5, label='Benign', color='blue' )
plt.hist(malignant_df["radius_mean"], bins = 40, alpha=0.5, label='Malignant', color='red')
plt.legend(loc='upper right')
plt.title('Distribution of Radius Mean: Malignant vs Benign')
plt.xlabel('Radius Mean (Variable)')
plt.ylabel('Count (Observations)')
plt.show()
  # Concavity Mean
plt.hist(benign_df["concavity_mean"], bins = 40, alpha = 0.5, label='Benign', color='blue' )
plt.hist(malignant_df["concavity_mean"], bins = 40, alpha=0.5, label='Malignant', color='red')
plt.legend(loc='upper right')
plt.title('Distribution of Concavity Mean: Malignant vs Benign')
plt.xlabel('Concavity Mean (Variable)')
plt.ylabel('Count (Observations)')
plt.show()

#Scatterplot (radius vs concavity)
plt.scatter(malignant_df['radius_mean'], malignant_df['concavity_mean'], color = 'red', alpha = 0.5, s = 8)
plt.scatter(benign_df['radius_mean'], benign_df['concavity_mean'], color = 'blue', alpha = 0.5, s = 8)
plt.legend(["Malignant", "Benign"], loc = "lower right")
plt.title('Malignant vs. Benign: Cell Radius and Concavity')
plt.xlabel('Radius Mean')
plt.ylabel('Concavity Mean')
plt.show()






```

Conclusions

I analyzed a tabular dataset of 569 patient observations to identify biological signals for malignancy. My analysis of the variables revealed that concavity_mean is the most significant indicator, with malignant cases averaging nearly 3.5 times higher than benign cases.

The Histograms confirm this strength, specifically showing a much smaller overlap in concavity distributions compared to other features, effectively isolating the majority of the benign population near zero. My Scatter Plot further proves a positive correlation between cell size and jaggedness in malignant cells. However, it is important to note that when the radius is near 15 and the concavity is near a value of 0.1, there is a 'grey zone' or overlap between malignant and benign datapoints. In this specific range, a definitive diagnosis is harder to make based on these two features alone, requiring further clinical investigation.
